Kalau datanya diunggah sebagai **zip** (boleh, disarankan satu zip per gitar), jalankan sel di bawah. Zip diekstrak ke disk lokal Colab (lebih cepat daripada membaca langsung dari Drive) dan `SRC` diarahkan ke sana. Kalau datanya sudah berupa folder biasa di Drive, lewati sel ini.

```python
import glob, os, subprocess

ZIP_DIR = '/content/drive/MyDrive/Data Train ZIP'   # folder Drive yang berisi file .zip
LOCAL = '/content/data_train'                        # disk lokal Colab (hilang saat sesi berakhir)

zips = sorted(glob.glob(os.path.join(ZIP_DIR, '*.zip')))
print(len(zips), 'zip:', [os.path.basename(z) for z in zips])
for z in zips:
    print('ekstrak', os.path.basename(z))
    subprocess.run(['unzip', '-q', '-o', z, '-d', LOCAL], check=True)

SRC = LOCAL
for name in sorted(os.listdir(LOCAL)):
    print(' ', name)
print('Kalau semua folder gitar ada di dalam satu folder lagi (misalnya "Data Train"), ubah SRC ke folder itu.')
```
