```python
# Contoh nama file di tiap folder (tempel keluaran sel ini ke chat Claude)
import os
for root, dirs, files in sorted(os.walk(DST)):
    wavs = sorted(f for f in files if f.lower().endswith('.wav'))
    if wavs:
        print(os.path.relpath(root, DST), '->', len(wavs), 'file:', ', '.join(wavs[:6]), '...' if len(wavs) > 6 else '')
```
